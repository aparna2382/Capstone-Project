```python
text_cols = ["department_of_production", "logistics_hub_cooperative",
             "trader_group", "country_of_destination"]
df_cocoa[text_cols] = df_cocoa[text_cols].apply(lambda s: s.str.strip().str.upper())

df_cocoa["commodity"] = "Cocoa"

# traceability gap
df_cocoa["is_unknown_origin"] = df_cocoa["department_of_production"].isin(
    ["UNKNOWN", "INDIRECT SOURCING"])

# EU flag
EU27 = ["AUSTRIA","BELGIUM","BULGARIA","CROATIA","CYPRUS","CZECHIA","CZECH REPUBLIC",
        "DENMARK","ESTONIA","FINLAND","FRANCE","GERMANY","GREECE","HUNGARY","IRELAND",
        "ITALY","LATVIA","LITHUANIA","LUXEMBOURG","MALTA","NETHERLANDS","POLAND",
        "PORTUGAL","ROMANIA","SLOVAKIA","SLOVENIA","SPAIN","SWEDEN"]
df_cocoa["is_eu"] = df_cocoa["country_of_destination"].isin(EU27)

# scan these lists for spelling differences and extra placeholder values
sorted(df_cocoa["country_of_destination"].unique())
```

```python
df_cocoa["eu_volume"] = df_cocoa["volume"].where(df_cocoa["is_eu"], 0)
df_cocoa["unknown_volume"] = df_cocoa["volume"].where(df_cocoa["is_unknown_origin"], 0)

check = df_cocoa.groupby("year").agg(
    rows=("volume", "size"),
    tonnes=("volume", "sum"),
    eu_tonnes=("eu_volume", "sum"),
    unknown_tonnes=("unknown_volume", "sum"),
    departments=("department_of_production", "nunique"),
    cooperatives=("logistics_hub_cooperative", "nunique"),
)
check["eu_share_%"] = (check["eu_tonnes"] / check["tonnes"] * 100).round(1)
check["unknown_origin_%"] = (check["unknown_tonnes"] / check["tonnes"] * 100).round(1)
check
```

```python
from pathlib import Path
Path("data/clean").mkdir(parents=True, exist_ok=True)
df_cocoa.to_csv("data/clean/cocoa_flows_clean.csv", index=False)
```

```python
d = df_cocoa_defor.drop(columns=["node_type_name", "node_type_slug", "country_slug"])
d = d.rename(columns={"region": "department",
                      "parent_region": "region_parent",
                      "numerical_value": "deforestation_ha"})

text_cols = ["department", "region_parent"]
d[text_cols] = d[text_cols].apply(lambda s: s.str.strip().str.upper())
d["deforestation_ha"] = pd.to_numeric(d["deforestation_ha"], errors="coerce")

print("duplicates:", d.duplicated(["department", "year"]).sum())   # must be 0
print("nulls:", d.isna().sum().sum())
d.groupby("year")["deforestation_ha"].agg(["count", "sum"]).round(0)
```

```python
flow_depts = set(df_cocoa["department_of_production"].unique())
defor_depts = set(d["department"].unique())

print("in flows, not in deforestation:", sorted(flow_depts - defor_depts))
print("in deforestation, not in flows:", len(defor_depts - flow_depts))
```
