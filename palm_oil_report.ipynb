```python
text_cols = ["province_of_production", "mill_group", "exporter", "country_of_first_import"]
df_palmoil[text_cols] = df_palmoil[text_cols].apply(lambda s: s.str.strip().str.upper())

df_palmoil["commodity"] = "Palm oil"

# traceability gap: mills Trase could not identify
df_palmoil["is_unknown_mill"] = df_palmoil["mill_group"].eq("UNKNOWN AFFILIATION")

# EU flag (needed for the EU-share KPI)
EU27 = ["AUSTRIA","BELGIUM","BULGARIA","CROATIA","CYPRUS","CZECHIA","CZECH REPUBLIC",
        "DENMARK","ESTONIA","FINLAND","FRANCE","GERMANY","GREECE","HUNGARY","IRELAND",
        "ITALY","LATVIA","LITHUANIA","LUXEMBOURG","MALTA","NETHERLANDS","POLAND",
        "PORTUGAL","ROMANIA","SLOVAKIA","SLOVENIA","SPAIN","SWEDEN"]
df_palmoil["is_eu"] = df_palmoil["country_of_first_import"].isin(EU27)

sorted(df_palmoil["country_of_first_import"].unique())   # scan this list for the missed EU countries
```

```python
df_palmoil["eu_volume"] = df_palmoil["volume"].where(df_palmoil["is_eu"], 0)
df_palmoil["unknown_volume"] = df_palmoil["volume"].where(df_palmoil["is_unknown_mill"], 0)

check = df_palmoil.groupby("year").agg(
    rows=("volume", "size"),
    tonnes=("volume", "sum"),
    eu_tonnes=("eu_volume", "sum"),
    unknown_mill_tonnes=("unknown_volume", "sum"),
)
check["eu_share_%"] = (check["eu_tonnes"] / check["tonnes"] * 100).round(1)
check["unknown_mill_%"] = (check["unknown_mill_tonnes"] / check["tonnes"] * 100).round(1)
check
```

```python
from pathlib import Path
Path("data/clean").mkdir(parents=True, exist_ok=True)
df_palmoil.to_csv("data/clean/palm_oil_flows_clean.csv", index=False)
```

```python
df_palm_defor=pd.read_csv("data/palm_oil/spatial-metrics-indonesia-territorial_deforestation_province.csv",index_col=0)
df_palm_defor = df_palm_defor.reset_index()
df_palm_defor
```

```python
p = df_palm_defor[df_palm_defor["year"].between(2020, 2024)].copy()
p = p.drop(columns=["country", "country_iso2", "parent_region_trase_id"])
p = p.rename(columns={"region": "province",
                      "parent_region": "island_group",
                      "deforestation_hectares": "deforestation_ha"})
p[["province", "island_group"]] = p[["province", "island_group"]].apply(lambda s: s.str.strip().str.upper())

print("rows per year:", p.groupby("year").size().to_dict())      # expect 38 provinces per year
print("duplicates:", p.duplicated(["province", "year"]).sum())   # must be 0
print("nulls:", p.isna().sum().sum())
print(p.groupby("year")["deforestation_ha"].sum().round(0))

# do the province names match the palm oil flow file?
flow_prov = set(df_palmoil["province_of_production"].unique())
defor_prov = set(p["province"].unique())
print("in flows, not in deforestation:", sorted(flow_prov - defor_prov))
print("in deforestation, not in flows:", sorted(defor_prov - flow_prov))
```

```python
p.to_csv("data/clean/palm_territorial_deforestation_clean.csv", index=False)
```
